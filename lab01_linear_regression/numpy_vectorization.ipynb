# NumPy 向量化
简要介绍 NumPy 向量化，并展示如何使用它来提高代码的性能。

## 创建向量

```python
import numpy as np
import time

# 分配内存并用数值填充数组的 NumPy 函数
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")

# 分配内存并用用户指定的数值填充的 NumPy 函数
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

## 向量运算

```python
#向量切片操作
a = np.arange(10)
print(f"a         = {a}")

#访问 5 个连续元素（start:stop:step）
c = a[2:7:1];     print("a[2:7:1] = ", c)

# 访问 3 个索引间隔为 2 的元素 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# 访问索引为 3 及以上的所有元素
c = a[3:];        print("a[3:]    = ", c)

# 访问索引小于 3 的所有元素
c = a[:3];        print("a[:3]    = ", c)

# 访问所有元素
c = a[:];         print("a[:]     = ", c)
```

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# 对 a 的元素取负
b = -a 
print(f"b = -a        : {b}")

# 对 a 的所有元素求和，返回一个标量
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")

# 将 a 乘以一个标量
b = 5 * a 
print(f"b = 5 * a : {b}")
```

向量与向量的逐元素运算
$$ c_i = a_i + b_i $$

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
```

当然，要使其正确执行，向量的大小必须相同

## 向量运算的高效 - 点积
$$ x = \sum_{i=0}^{n-1} a_i b_i $$

```python
# 实现自己的点积
def my_dot(a, b): 
    """
   计算两个向量的点积
 
    参数：
      a (ndarray (n,)):  输入向量 
      b (ndarray (n,)):  与 a 维数相同的输入向量
    
    返回：
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x


# 测试一维情况
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
```

```python
# 使用 numpy 提供的
print(f"np.dot(a, b) = {np.dot(a, b)}")
```

结果一致，看看效率

```python
np.random.seed(1)
a = np.random.rand(10000000)  # 非常大的数组
b = np.random.rand(10000000)

tic = time.time()  # 记录开始时间
c = np.dot(a, b)
toc = time.time()  # 记录结束时间

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # 记录开始时间
c = my_dot(a,b)
toc = time.time()  # 记录结束时间

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #从内存中移除这些大数组
```

很明显，向量化大幅提升了速度，这是因为 numpy 底层使用 C语言 实现，更加充分地利用了底层硬件中可用的数据并行性
GPU和现代CPU实现了单指令多数据(SIMD)流水线，允许并行发出多个操作
